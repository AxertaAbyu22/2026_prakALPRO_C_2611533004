## 2.1 Dasar Teori
### 1. Struktur Kontrol Perulangan Terarah (For Loop)
Perulangan `for` merupakan mekanisme kontrol alur yang dirancang untuk mengeksekusi blok kode secara repetitif berdasarkan barisan data yang telah ditentukan (*sequence* seperti *list*, *tuple*, maupun *string*), atau berdasarkan batasan numerik tertentu. Karakteristik utama yang membedakan `for-loop` dengan `while-loop` terletak pada kepastian jumlah iterasinya (*definite loop*). Jika `while-loop` bergantung dinamis pada evaluasi kondisi Boolean yang dapat memicu *infinite loop*, `for-loop` bekerja di dalam batas-batas indeks yang terukur sehingga titik akhir eksekusinya jauh lebih terprediksi sejak awal perulangan dimulai.

### 2. Pemanfaatan Fungsi range() Sebagai Generator Indeks
Dalam memproses perulangan berbasis angka, fungsi `range()` bertindak sebagai generator deret bilangan bulat secara berurutan. Fungsi ini memiliki arsitektur dasar `range(start, stop, step)`. Salah satu aspek krusial dalam implementasinya adalah sifat eksklusif dari parameter `stop`, di mana batas atas tersebut tidak akan pernah diikutsertakan ke dalam iterasi. Oleh karena itu, apabila program membutuhkan eksekusi tepat sebanyak N kali dengan indeks yang menyentuh angka N, nilai parameter pembatas yang dilewatkan ke dalam fungsi harus disesuaikan menjadi N + 1.

### 3. Logika Perulangan Bersarang (Nested Loop) Dua Dimensi
Perulangan bersarang (*nested loop*) terbentuk ketika sebuah struktur kontrol perulangan ditempatkan di dalam blok eksekusi perulangan lainnya. Dalam representasi data dua dimensi, kombinasi ini membagi peran secara terstruktur:
* **Outer Loop (Perulangan Luar):** Berfungsi sebagai pengendali dimensi vertikal atau perpindahan baris. Setiap satu kali iterasi pada *outer loop* selesai berjalan, program akan melakukan perpindahan ke baris baru di bawahnya.
* **Inner Loop (Perulangan Dalam):** Berfungsi sebagai pengendali dimensi horizontal. *Inner loop* akan menghabiskan seluruh jatah iterasinya terlebih dahulu untuk mencetak karakter demi karakter atau angka demi angka ke arah samping, sebelum kendali dikembalikan ke *outer loop*.

### 2.2.3 Program "perulangan_for3_2611533004.py"

```python
ulang_3004=int(input("masukan jumlaah perulangan :"))

jumlah_3004 = 0
for i_3004 in range(1, ulang_3004 + 1):
    print(i_3004, end=" ")
    jumlah_3004 = jumlah_3004 + i_3004

    if i_3004 < ulang_3004:
        print (" + ", end="")
    else:
        print (" = ", jumlah_3004, end="")
print()
print(" jumlah = ", jumlah_3004)
```

Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun. Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).

**Kode Jawaban :**

a. Kode Program "Tugas5_2611533004.py"

```python
# === PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

# Input dinamis ukuran skala jam pasir
n_3004 = int(input("Masukkan ukuran skala jam pasir (N): "))

# 1. BINGKAI PEMBATAS ATAS
print("#", end="")
for i_3004 in range(4 * n_3004 + 5):
    print("=", end="")
print("#")

# 2. FASE 1: JAM PASIR ATAS (Dari baris N turun sampai 1)
for baris_3004 in range(n_3004, 0, -1):
    print("| ", end="")  # Sisi kiri: garis tegak dan satu spasi padding
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_3004 in range(2 * (n_3004 - baris_3004)):
        print(" ", end="")
        
    # Deret angka mundur dari baris ke 1 dipisahkan spasi
    for angka_3004 in range(baris_3004, 0, -1):
        print(angka_3004, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 ke baris diawali spasi
    for angka_3004 in range(1, baris_3004 + 1):
        print(" " + str(angka_3004), end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_3004 in range(2 * (n_3004 - baris_3004)):
        print(" ", end="")
        
    print(" |")  # Satu spasi padding dan garis tegak di sisi kanan

# 3. FASE 2: POROS TITIK PUSAT (Singularity)
print("|", end="")  # Garis tegak pembatas kiri
# Spasi penyeimbang kiri: 2 * N + 1
for spasi_3004 in range(2 * n_3004 + 1):
    print(" ", end="")
print("<*>", end="")  # Poros kristal tunggal
# Spasi penyeimbang kanan: 2 * N + 1
for spasi_3004 in range(2 * n_3004 + 1):
    print(" ", end="")
print("|")  # Garis tegak pembatas kanan

# 4. FASE 3: JAM PASIR BAWAH (Dari baris 1 naik sampai N)
for baris_3004 in range(1, n_3004 + 1):
    print("| ", end="")  # Sisi kiri: garis tegak dan satu spasi padding
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_3004 in range(2 * (n_3004 - baris_3004)):
        print(" ", end="")
        
    # Deret angka mundur dari baris ke 1 dipisahkan spasi
    for angka_3004 in range(baris_3004, 0, -1):
        print(angka_3004, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 ke baris diawali spasi
    for angka_3004 in range(1, baris_3004 + 1):
        print(" " + str(angka_3004), end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_3004 in range(2 * (n_3004 - baris_3004)):
        print(" ", end="")
        
    print(" |")  # Satu spasi padding dan garis tegak di sisi kanan

# 5. BINGKAI PEMBATAS BAWAH
print("#", end="")
for i_3004 in range(4 * n_3004 + 5):
    print("=", end="")
print("#")
```

### Perancangan Bingkai Luar (Border)
Program membentuk bingkai pembatas geometris simetris menggunakan kalkulasi berbasis masukan nilai `n_3004`. Karakter pagar sudut (`#`) diletakkan di setiap ujung horizontal, mengapit barisan karakter garis ganda (`=`) yang diiterasi dinamis menggunakan formula `4 * n_3004 + 5`. Karakter dinding vertikal (`|`) dicetak pada sisi paling kiri dan kanan setiap baris untuk mengunci pola bagian dalam agar tampak terbungkus rapi.

### Struktur Jam Pasir Bagian Atas (Fase 1 - Reduksi)
Fase ini diatur oleh perulangan luar (*outer loop*) yang bergerak mundur dari batas `n_3004` turun sampai angka 1. Di dalam setiap barisnya, terdapat rangkaian perulangan dalam (*inner loop*) yang bekerja menyamping untuk mencetak spasi penyeimbang kiri dengan formula `2 * (n_3004 - baris_3004)`. Setelah spasi terbentuk, program mencetak deret angka mundur secara palindromik dari indeks baris aktif menuju angka 1. Bagian tengah baris dikunci oleh poros kristal tetap berupa karakter `<*>`, yang kemudian langsung disambung oleh deret angka maju dari angka 1 naik kembali menuju nilai indeks baris tersebut.

### Struktur Titik Pusat (Fase 2 - Singularity)
Titik pusat jam pasir merupakan fase tunggal yang memisahkan bagian atas dan bawah. Blok kode pada fase ini tidak bergantung pada perulangan luar yang dinamis, melainkan langsung mengeksekusi perulangan spasi penyeimbang tetap sebanyak `2 * n_3004 + 1` di sisi kiri dan kanan untuk menempatkan poros kristal tunggal `<*>` tepat berada di tengah struktur bangunan secara presisi.

### Struktur Jam Pasir Bagian Bawah (Fase 3 - Ekspansi)
Fase ini merepresentasikan cerminan vertikal dari Fase 1. Perulangan luar (*outer loop*) dikondisikan bergerak maju mulai dari angka 1 naik hingga mencapai batas atas `n_3004`. Seluruh logika *inner loop* yang mencetak deret spasi penyeimbang serta deret angka palindromik (mundur dan maju) memiliki arsitektur formula yang sama persis dengan Fase 1, namun menghasilkan output geometri yang bergerak melebar ke luar seiring bertambahnya nilai baris.

# KESIMPULAN

1. Berdasarkan hasil praktikum Pekan 5, dapat disimpulkan bahwa perulangan `for` dalam Python merupakan struktur kontrol alur terukur (*definite loop*) yang sangat efektif untuk mengeksekusi blok kode secara repetitif berdasarkan barisan objek data (*sequence*) maupun jangkauan indeks numerik yang pasti melalui fungsi `range()`.
2. Penggunaan arsitektur perulangan bersarang (*nested loop*) murni dengan mengombinasikan *outer loop* sebagai pengendali dimensi vertikal (baris) dan *inner loop* sebagai pengendali dimensi horizontal (kolom/karakter) merupakan pondasi fundamental dalam memetakan serta memanipulasi koordinat matriks dua dimensi.
3. Pembentukan pola geometri yang kompleks dan presisi, seperti Jam Pasir Kristal Palindromik Berbingkai, sangat bergantung pada integrasi formula matematis dinamis di dalam parameter indeks *looping*, yang memungkinan pengaturan spasi kosong (*padding*) dan susunan angka palindromik tercetak secara simetris tanpa menggunakan manipulasi perkalian string secara instan.
