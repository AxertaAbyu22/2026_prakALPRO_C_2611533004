## 2.1 Dasar Teori
Secara fundamental, percabangan dalam bahasa pemrograman Python merupakan suatu struktur kendali alur yang memungkinkan program untuk mengambil keputusan secara dinamis berdasarkan kondisi tertentu saat dijalankan. Penguasaan aspek percabangan menjadi basis utama dalam membangun alur algoritma yang cerdas serta mampu merespons berbagai logika program yang kompleks. Python menyediakan beberapa mekanisme dan komponen pendukung dalam mengimplementasikan logika kondisional ini, yaitu sebagai berikut:

1. Struktur Percabangan Dasar (If Tunggal dan If-Else):
Merupakan implementasi kondisional paling mendasar di mana If Tunggal berfungsi mengeksekusi blok kode tertentu hanya jika syarat terpenuhi. Sementara itu, struktur If-Else menyediakan jalur alternatif dua arah, sehingga program secara otomatis mengeksekusi blok kode di bawah klausul else apabila kondisi utama bernilai False.

2. Struktur Multi-Kondisi (If-Elif-Else):
Digunakan untuk menangani banyak skenario alternatif secara berurutan yang bersifat saling lepas (mutually exclusive). Sistem akan mengevaluasi kondisi dari atas ke bawah, dan proses pengecekan akan langsung dihentikan begitu salah satu blok kondisi bernilai True ditemukan.

3. Struktur Pencocokan Pola (Match-Case):
Fitur structural pattern matching yang diimplementasikan pada Python versi 3.10 ke atas sebagai alternatif yang lebih efisien dari struktur If-Elif-Else. Mekanisme ini dirancang khusus untuk mengoptimalkan efisiensi penulisan kode saat melakukan pengujian terhadap opsi-opsi nilai yang bersifat ekuivalen.

4. Kondisi Majemuk (Ekspresi Logika):
Instrumen untuk merancang alur logika kompleks dengan mengevaluasi beberapa syarat sekaligus dalam satu baris instruksi tanpa harus menerapkan percabangan bersarang (nested-if). Implementasinya memanfaatkan operator logika and (seluruh syarat harus terpenuhi), or (minimal satu syarat terpenuhi), serta operator relasional ketidaksamaan (!=).

5. Multi-if Independen:
Rangkaian instruksi percabangan yang berdiri sendiri secara sejajar dan terpisah. Berbeda dengan karakteristik If-Elif, setiap pernyataan dalam struktur Multi-if independen akan selalu dievaluasi secara mandiri dari atas ke bawah oleh sistem, menjadikannya metode wajib untuk menangani operasi yang bersifat akumulatif.

### 2.2.5 Program "multi_if2_2611533004.py"

```python
# Buat file dengan nama multi_if2_2611533004.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit NIM terakhir contoh: total_belanja_3004
# program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3004 = float(input("Input Total Belanja Anda (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3004 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_3004 = input_member_3004 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3004 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3004 = input_promo_3004 in ["y", "ya"]

total_diskon_persen_3004 = 0

# Multi-IF terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3004 > 1000000:
    total_diskon_persen_3004 += 10 # Diskon belanja besar

if is_member_3004:
    total_diskon_persen_3004 += 5 # Diskon member

if kode_promo_valid_3004:
    total_diskon_persen_3004 += 15 # Diskon voucher

# menghitung nominal diskon dan total bayar
nominal_diskon_3004 = total_belanja_3004 * (total_diskon_persen_3004 / 100)
total_bayar_3004 = total_belanja_3004 - nominal_diskon_3004

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_3004}% (Rp {nominal_diskon_3004:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_3004:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3004}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid

# Output : total diskon yang anda dapatkan : 30% jika belanja > 1 juta, member, dan kode promo valid
```

### 2.2.7 Program "match_case1_2611533004.py"

```python
bulan_3004 =int(input("Masukan Bulan (1-12): "))

match bulan_3004:
    case 1:
        print("Januari")
    case 2:
        print("februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("July")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak Valiid")

print("Program Selesai")
```

a. Kode Program "Tugas4_2611533004.py"

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")
nama_3004 = input("Masukkan Nama Pengunjung        : ")
umur_3004 = int(input("Input umur anda                 : "))

# Mengambil huruf pertama dari input SIM untuk validasi yang akurat
sim_input_3004 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()
sim_3004 = sim_input_3004[0] if len(sim_input_3004) > 0 else 't'

print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_3004 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_3004 = int(input("Masukkan jumlah tiket           : "))
member_3004 = input("Apakah Anda member? (y/t)       : ").strip().lower()
promo_3004 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# Variabel flag untuk menghentikan eksekusi tanpa exit() atau nested-if
validasi_3004 = True

# If Tunggal untuk validasi kelogisan tiket
if jumlah_tiket_3004 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")
    validasi_3004 = False

harga_satuan_3004 = 0
nama_wahana_3004 = ""

# Match - Case beserta Case _
match paket_3004:
    case 1:
        nama_wahana_3004 = "Safari Rimba"
        harga_satuan_3004 = 50000
    case 2:
        nama_wahana_3004 = "Arung Jeram"
        harga_satuan_3004 = 75000
    case 3:
        nama_wahana_3004 = "Motor ATV Ekstrim"
        harga_satuan_3004 = 120000
    case 4:
        nama_wahana_3004 = "Roller Coaster Kilat"
        harga_satuan_3004 = 100000
    case 5:
        nama_wahana_3004 = "All-Access VIP"
        harga_satuan_3004 = 220000
    case _:
        print("Paket wahana tidak valid!")
        validasi_3004 = False

# Jika validasi_3004 bernilai False (tiket <= 0 atau paket tidak valid), 
# maka program akan melewati seluruh evaluasi di bawah ini secara otomatis.

if validasi_3004:
    print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

# If - Elif - Else terstruktur tanpa Nested If (if di dalam if)
if validasi_3004 and paket_3004 == 3 and umur_3004 >= 17 and sim_3004 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif validasi_3004 and paket_3004 == 3 and umur_3004 >= 17 and sim_3004 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh membawa motor ATV (wajib didampingi instruktur).")
elif validasi_3004 and paket_3004 == 3 and umur_3004 < 17 and sim_3004 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif validasi_3004 and paket_3004 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh membawa motor ATV.")
elif validasi_3004 and umur_3004 >= 10:
    print(f"Status Akses: Umur Anda mencukupi untuk wahana {nama_wahana_3004}.")
elif validasi_3004:
    print(f"Status Akses: Anda belum cukup umur untuk wahana {nama_wahana_3004} (wajib didampingi orang tua).")

# Multi-IF terpisah untuk diskon akumulatif
subtotal_3004 = harga_satuan_3004 * jumlah_tiket_3004
total_diskon_persen_3004 = 0

if validasi_3004 and subtotal_3004 >= 200000:
    total_diskon_persen_3004 += 10
if validasi_3004 and member_3004 in ['y', 'ya']:
    total_diskon_persen_3004 += 5
if validasi_3004 and promo_3004 in ['y', 'ya']:
    total_diskon_persen_3004 += 15
if validasi_3004 and jumlah_tiket_3004 >= 5:
    total_diskon_persen_3004 += 5

# Kalkulasi Nominal Diskon dan Evaluasi Akhir
if validasi_3004:
    nominal_diskon_3004 = subtotal_3004 * (total_diskon_persen_3004 / 100)
    total_bayar_3004 = subtotal_3004 - nominal_diskon_3004

    print("\n--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {subtotal_3004:,.0f}")
    print(f"Total Diskon     : {total_diskon_persen_3004}% (Rp {nominal_diskon_3004:,.0f})")
    print(f"Total Bayar      : Rp {total_bayar_3004:,.0f}")

# Evaluasi if - else penutup audit
if validasi_3004 and total_bayar_3004 > 300000:
    print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
elif validasi_3004:
    print("Catatan Layanan  : Terima kasih telah berkunjung.")
```

Karena nomor paket yang dimasukkan adalah 5, alur program diarahkan masuk ke dalam case 5 yang secara otomatis menetapkan string variabel nama wahana menjadi "All-Access VIP" dan mengunci variabel harga satuan tiket pada angka Rp 220.000. Selama jumlah tiket yang dimasukkan bernilai valid atau logis (> 0), variabel flag validasi akan tetap terkunci pada status boolean True.

### Tahap Evaluasi Kelayakan Akses (If-Elif-Else)
Program melakukan penyaringan terhadap hak kelayakan pengendara menggunakan struktur percabangan if-elif-else terstruktur tanpa nested-if. Dikarenakan paket yang dibeli adalah paket 5 (bukan paket 3/ATV) dan nilai variabel umur bernilai 18 tahun, sistem melompati pengecekan prasyarat SIM C dan langsung mencocokkan kesesuaian pada kondisi syarat umur >= 10 tahun. Hasil evaluasi ini memicu status akses valid yang menyatakan pengunjung diizinkan menikmati wahana tersebut.

### Tahap Perhitungan Diskon Akumulatif (Multi-If Independen)
Setelah harga tiket dan kuota divalidasi, program menghitung nilai subtotal belanja menggunakan operator aritmatika perkalian (*) hingga didapatkan angka Rp 880.000. Nilai persentase potongan harga kemudian diperiksa menggunakan rangkaian struktur multi-if terpisah yang bekerja secara independen dari atas ke bawah. Karena subtotal melampaui Rp 200.000 (+10%), pengunjung berstatus member (+5%), dan kode promo dikonfirmasi valid (+15%), program berhasil mengumpulkan total persentase diskon secara akumulatif sebesar 30%. Persentase ini kemudian dikalkulasikan untuk menghasilkan total pembayaran bersih.

### Tahap Audit Penutup Total Transaksi (If-Else)

Pada tahap akhir eksekusi, program melakukan fungsi audit penutup terhadap nominal transaksi kasir menggunakan struktur percabangan if-else. Karena nominal pada variabel total bayar akhir menyentuh angka Rp 616.000, di mana nilai tersebut terbukti jauh lebih besar dari parameter prasyarat minimum Rp 300.000, program secara otomatis mengaktifkan blok pengkondisian pertama untuk mencetak struk bonus souvenir gratis kepada pelanggan.
