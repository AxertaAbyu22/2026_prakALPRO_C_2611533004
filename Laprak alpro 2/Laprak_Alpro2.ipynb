## 2.1 Dasar Teori
1. Tipe Data Numerik: Python memfasilitasi komputasi numerik melalui penyediaan tipe data bilangan bulat (int), bilangan desimal bertipe pecahan (float), serta bilangan kompleks (complex) guna mengakomodasi berbagai kebutuhan operasi matematika dan kalkulasi algoritma.
2. Tipe Data String: Tipe data ini digunakan untuk merepresentasikan dan mengolah sekumpulan karakter atau teks. Pembuatannya dapat diapit oleh tanda petik tunggal ('), ganda ("), maupun tiga kali lipat (""") untuk menampung teks yang terdiri dari beberapa baris sekaligus (multiline).
3. Tipe Data Boolean: Merupakan tipe data logika fundamental yang hanya merepresentasikan dua nilai kebenaran, yaitu True (benar) dan False (salah). Tipe ini memegang peranan krusial dalam mengontrol alur eksekusi pada struktur percabangan program.
4. Input Data Dinamis: Berbeda dari metode statis yang nilainya bersifat kaku di dalam kode, fungsi input() memberikan fleksibilitas pada sistem untuk menerima data secara interaktif dari pengguna melalui terminal selama program berjalan.
5. Konstanta: Kendati Python secara bawaan tidak mengunci nilai variabel, implementasi konstanta tetap diterapkan untuk data yang bersifat absolut dan permanen. Praktik ini umumnya dipertegas melalui penulisan huruf kapital atau penggunaan modul dekorator Final.


## 2.2 Kode Program

### 2.2.1 Program "Numerik_2611533004.py"
Program berikut mengilustrasikan penciptaan beberapa variabel, masing-masing dengan muatan tipe numerik yang berbeda (integer, float, complex), kemudian mengidentifikasi validasi struktur bawaannya memakai built-in function `type()`.

```python
# Buat file dengan nama Numerik_NIM.py
# nama variable ditambah 4 digit terakhir NIM contoh:a_3004
#create a variable with numeric type value.

a_3004=100
print("The type of variable having value", a_3004, "is", type(a_3004))

#create a variable with float type value.
b_3004=10.2345
print("The type of variable having value", b_3004, "is", type (b_3004))

#create a variable with complex type value.
c_3004=100+3j
print("The type of variable having value", c_3004, "is", type (c_3004))
```

**Penjelasan :**
1. `a_3004 = 100` : Berfungsi untuk mendeklarasikan variabel `a_3004` dan menginisialisasinya dengan nilai bilangan bulat (integer), yaitu `100`.
2. `b_3004 = 10.23455` : Berfungsi untuk mendeklarasikan variabel `b_3004` dengan menugaskan nilai berupa pecahan atau desimal (float).
3. `c_3004 = 100+3j` : Berfungsi untuk mendeklarasikan variabel `c_3004` dengan nilai bilangan kompleks (complex) bawaan Python yang ditandai dengan sufiks `j`.
4. `type(...)` pada fungsi `print` : Berfungsi sebagai *built-in function* dari Python yang digunakan untuk mengecek tipe data aktual dari variabel yang ada di dalam tanda kurungnya, dan mencetak nama kelas tipe datanya ke layar terminal.

### 2.2.2 Program String "String_2611533004"
Blok program identitas diri di bawah ini mengandalkan string untuk mendefinisikan informasi profil. Teks dapat ditampung dalam sebaris tanda petik maupun kutipan multiline string ketika mendefinisikan lokasi alamat secara berderet.

```python
nama_3004 = "Abiyyu"
jenis_kelamin_3004= 'L'
alamat_3004="""
    jalan belibis blok a no 5
    Kec.padang utara
    Kota Padang
"""
agama_3004 = 'islam'
print("Nama saya adalah: ", nama_3004)
print(" Jenis kelamin saya adalah:", jenis_kelamin_3004)
print(" Alamat saya adalah:", alamat_3004)
print(" Agama saya adalah:", agama_3004)
```

Penjelasan program driver di atas: ...

1. `nama_3004 = "Abiyyu"` dan `jenis_kelamin_3004 = 'L'` : Berfungsi menginisiasi variabel tipe data teks (string). Python mengizinkan pemakaian tanda kutip dua (`" "`) maupun tanda kutip tunggal (`' '`) untuk membungkus sebaris teks.
2. `alamat_3004 = """..."""` : Berfungsi untuk menampung teks yang berisi lebih dari satu baris atau *multiline string*. Teks dibungkus menggunakan tanda kutip tiga (bisa `"""` atau `'''`) agar perpindahan baris (enter) ikut tersimpan sebagai bagian dari data alamat tersebut.
3. `print("Nama saya adalah:", nama_3004)` : Berfungsi untuk mencetak gabungan string karakter statis dengan nilai yang tersimpan di dalam suatu variabel string.

### 2.2.3 Program Boolean "Boolean_2611533004"
Program berikut mengaplikasikan boolean untuk mengecek syarat parameter nilai kelulusan. Status nilai akan dicari kepastiannya (`True` atau `False`) sebelum divalidasi dan diizinkan melaju memanggil perintah cetak Cum Laude.

```python
is_lulus_3004 = True
is_cumlaude_3004 = True

# menggunakan Boolean
nilai_3004= 85
batas_lulus_3004= 75

# menggunakan nilai Boolean dari kondisi
status_kelulusan_3004 = nilai_3004 >= batas_lulus_3004 # Hasilnya akan True

print("=== cek kelulusan ===")
print("Nilai:", nilai_3004)
print("Apakah lulus?:", status_kelulusan_3004)
if is_lulus_3004 and is_cumlaude_3004:
    print("Selamat, anda lulus dengan predikat cumlaude!")
```

Penjelasan program boolean di atas:

1. `is_lulus_3004 = True` dan `is_cumlaude_3004 = True` : Berfungsi mendeklarasikan variabel tipe data Boolean yang merepresentasikan kebenaran, di mana nilainya hanya dapat diatur sebagai `True` (benar) atau `False` (salah).
2. `status_kelulusan_3004 = nilai_3004 >= batas_lulus_3004` : Berfungsi melakukan evaluasi operator perbandingan relasional (lebih besar sama dengan). Karena nilai 85 lebih besar dari 75, hasil evaluasi tersebut akan mengembalikan nilai kebenaran Boolean berupa `True` lalu disalurkan ke variabel `status_kelulusan_3004`.
3. `if is_lulus_3004 and is_cumlaude_3004:` : Berfungsi sebagai blok pengkondisian menggunakan operator logika `and`. Program hanya akan mengeksekusi fungsi print berupa ucapan "Selamat!" yang ada di bawahnya apabila kondisi di kiri dan kanan sintaks `and` sama-sama bernilai `True`.

### 2.2.4 Program Input "Input_2611533004"
Skrip program ini bergantung pada `input()` agar sisi pengerjaan perumusan dapat secara fleksibel disuplai masukan dari sisi end-user. Masukan awal direkam menggunakan string default untuk lalu dipaksa berubah bentuk (`type casting`) menjadi bilangan `int` agar kalkulasi dimensi berjalan lancar.

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variable ditambah 4 digit terakhir NIM contoh:panjang_3004, lebar_3004, luas_3004
# Program menghitung luas persegi panjang
# memasukkan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan di konversikan oleh output
panjang_3004=input("masukkan panjang: ")
lebar_3004=input("masukkan lebar: ")
luas_3004= int(panjang_3004) * int(lebar_3004)
print("luas persegi panjang= ", luas_3004)
```

**Penjelasan :**
1. `input('Masukkan nilai panjang: ')` : Berfungsi menampilkan teks sapaan ke layar, kemudian menahan eksekusi program sejenak untuk menunggu pengguna mengetikkan sesuatu (inputan) ke dalam *console* / *terminal*. Secara bawaan (default), semua masukan akan direkam sebagai data String.
2. `int(panjang_3004)` dan `int(lebar_3004)` : Berfungsi melakukan proses *type casting* atau pemaksaan konversi tipe data.[cite: 2] Nilai panjang dan lebar yang awalnya berupa karakter teks atau String diubah paksa menjadi bilangan bulat (Integer) agar dapat dikalikan dalam operasi matematika `*` (perkalian).
3. `luas_3004 = ...` : Menyimpan hasil akhir nilai perkalian yang bertipe Integer.

### 2.2.5 Program Konstanta "Konstanta_2611533004"
Menggunakan import atribut `Final` dari modul `typing`, kode ini mempatenkan variabel global `PI` senilai 3.14. Tujuannya adalah mereduksi kesalahan perubahan nilai sembarang secara tidak sengaja ketika program dikomputasi untuk menelusuri luas lingkaran.

```python
# Buat file dengan nama konstanta_2611533004.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran.
# Nama variabel ditambah 4 digit nim terakhir contoh: jari_3004

from typing import final
PI: final = 3.1415
print("pi: %f" % (PI))
jari_3004 =float(input("masukkan nilai jari-jari: "))
luas_3004 = PI * jari_3004 * jari_3004
print("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3004, luas_3004))
```

**Penjelasan :**
1. `from typing import Final` : Berfungsi untuk memanggil library atau modul bawaan `typing` dari Python, guna menggunakan kemampuan pelabelan (anotasi) jenis variabel tertentu dalam hal ini sebagai tipe `Final`.
2. `PI: Final = 3.14` : Mendeklarasikan bahwa variabel `PI` bersifat final atau konstan. Ini menunjukkan pada struktur kode bahwa `PI` merupakan tetapan nilai matematis yang tidak boleh diubah oleh baris kode mana pun secara tidak sengaja di masa mendatang.
3. `float(input(...))` : Meminta input dari pengguna, namun berkat pembungkusan fungsi tipe data `float(...)`, teks angka yang dimasukkan pengguna secara sekejap langsung dikonversi ke tipe bilangan desimal (float).
4. `%f` dan `%.2f` : Berfungsi sebagai teknik pemformatan string kuno gaya bahasa C (*String Formatting*). Simbol `%f` mencetak bilangan float secara utuh, sedangkan `%.2f` digunakan untuk membatasi tampilan hasil hitung luas dan input jari-jari agar hanya menyisakan dua digit angka spesifik saja setelah tanda koma.s
