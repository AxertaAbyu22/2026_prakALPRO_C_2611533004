# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori
Untuk membangun fondasi yang kuat dalam rekayasa perangkat lunak, pemahaman terhadap alat kerja (*tooling*) dan bahasa pemrograman yang digunakan sangatlah krusial. Berikut adalah penjelasan mengenai tiga komponen utama yang digunakan dalam praktikum ini:

### 2.1.1 Bahasa Pemrograman Python
Python merupakan bahasa pemrograman tingkat tinggi yang dirancang dengan filosofi keterbacaan kode (*code readability*). Berbeda dengan bahasa pemrograman lain yang penuh dengan simbol kaku, Python menggunakan sintaks yang bersih dan cenderung mirip dengan bahasa Inggris operasional sehari-hari. 

Keunggulan utama Python terletak pada kesederhanaannya tanpa mengorbankan fungsionalitas. Python mendukung berbagai paradigma pemrograman—seperti prosedural dan berorientasi objek—serta didukung oleh ekosistem pustaka (*library*) yang sangat luas. Hal ini menjadikannya pilihan utama di berbagai bidang modern, mulai dari pengembangan web, otomatisasi sistem, hingga kecerdasan buatan (*Artificial Intelligence*).

### 2.1.2 Visual Studio Code (VS Code)
Menulis kode yang efisien membutuhkan lingkungan kerja yang suportif. Visual Studio Code atau VS Code hadir sebagai *code editor* lintas platform yang ringan namun sangat bertenaga. Dikembangkan oleh Microsoft, VS Code tidak hanya berfungsi sebagai tempat mengetik teks biasa, melainkan sebuah ekosistem yang dapat disesuaikan.

Melalui dukungan *extension store* yang kaya, VS Code dapat diubah menjadi *Integrated Development Environment* (IDE) mini yang canggih. Fitur-fitur seperti *IntelliSense* (auto-complete kode), terminal terintegrasi, serta fitur manajemen Git bawaan membuat proses *debugging* dan penulisan skrip Python menjadi jauh lebih cepat dan terstruktur.

### 2.1.3 Sistem Pengontrol Versi (Git & GitHub)
Dalam dunia pengembangan perangkat lunak, perubahan kode terjadi hampir setiap menit. Di sinilah pentingnya *Version Control System* (VCS) seperti Git. Git bekerja seperti "mesin waktu" yang mencatat setiap riwayat perubahan, penambahan, atau penghapusan baris kode secara lokal di komputer pengembang melalui perintah *commit*.

Sementara itu, GitHub bertindak sebagai platform berbasis *cloud* tempat menyimpan dan membagikan repositori Git tersebut secara *online*. Kolaborasi antara Git dan GitHub memungkinkan pengembang untuk mengamankan cadangan kode mereka, bekerja bersama tim tanpa takut kodenya saling tertimpa, serta melacak perkembangan proyek dari waktu ke waktu secara transparan.


## 2.2 Kode Program

### 2.2.1 Program "program1_2611533004"
Berikut adalah implementasi kode program pertama menggunakan sintaks dasar Python.

```python
# Program pertama
"""
komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Departemen Informatika")
print("Universitas Andalas")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini"+ " dibuat oleh a\\n"+ "Muhammad Abiyyu Haidar Arhab_2611533004")
```

**Penjelasan :**
1. **`""" ... """` (Docstring)**: Bagian pertama kode ini menggunakan tanda kutip tiga untuk membuat komentar multi-baris (*multiline comment*). Baris ini tidak akan dieksekusi oleh program dan berfungsi sebagai catatan.
2. **`print()`**: Fungsi dasar untuk mencetak teks ke layar. `print()` yang dibiarkan kosong akan mencetak baris kosong (sebagai jarak).
3. Karakter Escape `\\`**: Untuk mencetak simbol *backslash* (`\`) di Python, kita harus menggunakan *double backslash* (`\\`) agar tidak dibaca sebagai awalan perintah khusus.
4. **Konkatenasi `+` dan `\n`**: Pada baris terakhir, digunakan simbol `+` untuk menggabungkan (*concatenate*) beberapa *string*. Terdapat juga `\n` (newline) yang berfungsi untuk memaksa teks setelahnya turun ke baris baru, sehingga nama Afnan Feyza dicetak di baris bawah.
